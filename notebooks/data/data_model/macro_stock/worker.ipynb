```python
import json

catalog = dbutils.widgets.get("catalog")
ticker = dbutils.widgets.get("ticker")
start_date = dbutils.widgets.get("start_date")
```

```python
df_stock = spark.table(f"{catalog}.yfinance.`{ticker}`").createOrReplaceTempView("df_stock")
df_macro = spark.table(f"{catalog}.fred.macro_indicators").createOrReplaceTempView("df_macro")
```

```python
queries = f"""
    CREATE OR REPLACE TEMPORARY VIEW joined as (
        select
            df_stock.*,
            df_macro.GDPC1, df_macro.CPIAUCSL, df_macro.UNRATE, df_macro.FEDFUNDS, df_macro.DFF,
            df_macro.DGS10, df_macro.DGS2, df_macro.T10Y2Y, df_macro.VIXCLS, df_macro.M2SL
        from df_stock
        left join df_macro
            on df_stock.date = df_macro.date
        where df_stock.date > date_sub('{start_date}',31)
    );
    
    CREATE OR REPLACE TEMPORARY VIEW bounds as (
        select
            date, open, high, low, close, adj_close, volume,
            GDPC1,
            MAX(
                CASE WHEN GDPC1 IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS GDPC1_x0,
            MAX(
                CASE WHEN GDPC1 IS NOT NULL 
                THEN GDPC1 END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS GDPC1_y0,
            MIN(
                CASE WHEN GDPC1 IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS GDPC1_x1,
            MIN(
                CASE WHEN GDPC1 IS NOT NULL 
                THEN GDPC1 END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS GDPC1_y1,
            CPIAUCSL,
            MAX(
                CASE WHEN CPIAUCSL IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS CPIAUCSL_x0,
            MAX(
                CASE WHEN CPIAUCSL IS NOT NULL 
                THEN CPIAUCSL END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS CPIAUCSL_y0,
            MIN(
                CASE WHEN CPIAUCSL IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS CPIAUCSL_x1,
            MIN(
                CASE WHEN CPIAUCSL IS NOT NULL 
                THEN CPIAUCSL END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS CPIAUCSL_y1,
            UNRATE,
            MAX(
                CASE WHEN UNRATE IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS UNRATE_x0,
            MAX(
                CASE WHEN UNRATE IS NOT NULL 
                THEN UNRATE END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS UNRATE_y0,
            MIN(
                CASE WHEN UNRATE IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS UNRATE_x1,
            MIN(
                CASE WHEN UNRATE IS NOT NULL 
                THEN UNRATE END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS UNRATE_y1,
            FEDFUNDS,
            MAX(
                CASE WHEN FEDFUNDS IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS FEDFUNDS_x0,
            MAX(
                CASE WHEN FEDFUNDS IS NOT NULL 
                THEN FEDFUNDS END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS FEDFUNDS_y0,
            MIN(
                CASE WHEN FEDFUNDS IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS FEDFUNDS_x1,
            MIN(
                CASE WHEN FEDFUNDS IS NOT NULL 
                THEN FEDFUNDS END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS FEDFUNDS_y1,
            DFF,
            MAX(
                CASE WHEN DFF IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS DFF_x0,
            MAX(
                CASE WHEN DFF IS NOT NULL 
                THEN DFF END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS DFF_y0,
            MIN(
                CASE WHEN DFF IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS DFF_x1,
            MIN(
                CASE WHEN DFF IS NOT NULL 
                THEN DFF END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS DFF_y1,
            DGS10,
            MAX(
                CASE WHEN DGS10 IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS DGS10_x0,
            MAX(
                CASE WHEN DGS10 IS NOT NULL 
                THEN DGS10 END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS DGS10_y0,
            MIN(
                CASE WHEN DGS10 IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS DGS10_x1,
            MIN(
                CASE WHEN DGS10 IS NOT NULL 
                THEN DGS10 END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS DGS10_y1,
            DGS2,
            MAX(
                CASE WHEN DGS2 IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS DGS2_x0,
            MAX(
                CASE WHEN DGS2 IS NOT NULL 
                THEN DGS2 END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS DGS2_y0,
            MIN(
                CASE WHEN DGS2 IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS DGS2_x1,
            MIN(
                CASE WHEN DGS2 IS NOT NULL 
                THEN DGS2 END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS DGS2_y1,
            T10Y2Y,
            MAX(
                CASE WHEN T10Y2Y IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS T10Y2Y_x0,
            MAX(
                CASE WHEN T10Y2Y IS NOT NULL 
                THEN T10Y2Y END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS T10Y2Y_y0,
            MIN(
                CASE WHEN T10Y2Y IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS T10Y2Y_x1,
            MIN(
                CASE WHEN T10Y2Y IS NOT NULL 
                THEN T10Y2Y END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS T10Y2Y_y1,
            VIXCLS,
            MAX(
                CASE WHEN VIXCLS IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS VIXCLS_x0,
            MAX(
                CASE WHEN VIXCLS IS NOT NULL 
                THEN VIXCLS END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS VIXCLS_y0,
            MIN(
                CASE WHEN VIXCLS IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS VIXCLS_x1,
            MIN(
                CASE WHEN VIXCLS IS NOT NULL 
                THEN VIXCLS END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS VIXCLS_y1,
            M2SL,
            MAX(
                CASE WHEN M2SL IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS M2SL_x0,
            MAX(
                CASE WHEN M2SL IS NOT NULL 
                THEN M2SL END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
            ) AS M2SL_y0,
            MIN(
                CASE WHEN M2SL IS NOT NULL 
                THEN date END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS M2SL_x1,
            MIN(
                CASE WHEN M2SL IS NOT NULL 
                THEN M2SL END
            ) OVER (
                ORDER BY date ROWS 
                BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
            ) AS M2SL_y1
        from joined
    );

    CREATE OR REPLACE TEMPORARY VIEW filled as (
        select
            date, open, high, low, close, adj_close, volume,
            COALESCE(
                GDPC1,
                GDPC1_y0 + datediff(date, GDPC1_x0) * (GDPC1_y1 - GDPC1_y0) / NULLIF(datediff(GDPC1_x1, GDPC1_x0), 0),
                GDPC1_y1,
                GDPC1_y0
            ) AS GDPC1,
            COALESCE(
                CPIAUCSL,
                CPIAUCSL_y0 + datediff(date, CPIAUCSL_x0) * (CPIAUCSL_y1 - CPIAUCSL_y0) / NULLIF(datediff(CPIAUCSL_x1, CPIAUCSL_x0), 0),
                CPIAUCSL_y1,
                CPIAUCSL_y0
            ) AS CPIAUCSL,
            COALESCE(
                UNRATE,
                UNRATE_y0 + datediff(date, UNRATE_x0) * (UNRATE_y1 - UNRATE_y0) / NULLIF(datediff(UNRATE_x1, UNRATE_x0), 0),
                UNRATE_y1,
                UNRATE_y0
            ) AS UNRATE,
            COALESCE(
                FEDFUNDS,
                FEDFUNDS_y0 + datediff(date, FEDFUNDS_x0) * (FEDFUNDS_y1 - FEDFUNDS_y0) / NULLIF(datediff(FEDFUNDS_x1, FEDFUNDS_x0), 0),
                FEDFUNDS_y1,
                FEDFUNDS_y0
            ) AS FEDFUNDS,
            COALESCE(
                DFF,
                DFF_y0 + datediff(date, DFF_x0) * (DFF_y1 - DFF_y0) / NULLIF(datediff(DFF_x1, DFF_x0), 0),
                DFF_y1,
                DFF_y0
            ) AS DFF,
            COALESCE(
                DGS10,
                DGS10_y0 + datediff(date, DGS10_x0) * (DGS10_y1 - DGS10_y0) / NULLIF(datediff(DGS10_x1, DGS10_x0), 0),
                DGS10_y1,
                DGS10_y0
            ) AS DGS10,
            COALESCE(
                DGS2,
                DGS2_y0 + datediff(date, DGS2_x0) * (DGS2_y1 - DGS2_y0) / NULLIF(datediff(DGS2_x1, DGS2_x0), 0),
                DGS2_y1,
                DGS2_y0
            ) AS DGS2,
            COALESCE(
                T10Y2Y,
                T10Y2Y_y0 + datediff(date, T10Y2Y_x0) * (T10Y2Y_y1 - T10Y2Y_y0) / NULLIF(datediff(T10Y2Y_x1, T10Y2Y_x0), 0),
                T10Y2Y_y1,
                T10Y2Y_y0
            ) AS T10Y2Y,
            COALESCE(
                VIXCLS,
                VIXCLS_y0 + datediff(date, VIXCLS_x0) * (VIXCLS_y1 - VIXCLS_y0) / NULLIF(datediff(VIXCLS_x1, VIXCLS_x0), 0),
                VIXCLS_y1,
                VIXCLS_y0
            ) AS VIXCLS,
            COALESCE(
                M2SL,
                M2SL_y0 + datediff(date, M2SL_x0) * (M2SL_y1 - M2SL_y0) / NULLIF(datediff(M2SL_x1, M2SL_x0), 0),
                M2SL_y1,
                M2SL_y0
            ) AS M2SL
        from bounds
    );
"""
for q in queries.split(";"):
    if q.strip() != "":
        spark.sql(q.strip())
```

```python
from delta.tables import DeltaTable

tbs = [t.table_name for t in spark.sql(f"SHOW TABLES IN {catalog}.yfinance")]

if f"{ticker}_enriched" in tbs:
    deltaTable = DeltaTable.forName(f"{catalog}.yfinance.`{ticker}_enriched`")

    deltaTable.alias("target").merge(
        spark.table("filled").alias("source"), 
        "target.date = source.date"
    ).whenMatchedUpdateAll()\
        .whenNotMatchedInsertAll()\
        .execute()
else:
    spark.table("filled")\
        .write.format("delta")\
        .mode("overwrite")\
        .saveAsTable(f"{catalog}.yfinance.{ticker}_enriched")

last_date = spark.sql("""
    select
        LEAST(
            MAX(CASE WHEN GDPC1 IS NOT NULL THEN date END),
            MAX(CASE WHEN CPIAUCSL IS NOT NULL THEN date END),
            MAX(CASE WHEN UNRATE IS NOT NULL THEN date END),
            MAX(CASE WHEN FEDFUNDS IS NOT NULL THEN date END),
            MAX(CASE WHEN DFF IS NOT NULL THEN date END),
            MAX(CASE WHEN DGS10 IS NOT NULL THEN date END),
            MAX(CASE WHEN DGS2 IS NOT NULL THEN date END),
            MAX(CASE WHEN T10Y2Y IS NOT NULL THEN date END),
            MAX(CASE WHEN VIXCLS IS NOT NULL THEN date END),
            MAX(CASE WHEN M2SL IS NOT NULL THEN date END)
        ) AS last_date
    from joined
""").collect()[0]["last_date"]

dbutils.notebook.exit(
    json.dumps(
        {
            "ingestionStatus": 1, 
            "last_date": str(last_date) if last_date else None
        }
    )
)
```

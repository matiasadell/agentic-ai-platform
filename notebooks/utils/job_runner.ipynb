```python
from databricks.sdk import WorkspaceClient
from databricks.sdk.service.jobs import RunResultState

dbutils.widgets.text("job_name", "")
job_name = dbutils.widgets.get("job_name")
```

```python
client = WorkspaceClient()
job = next(client.jobs.list(name=job_name))

run = client.jobs.run_now(job_id=job.job_id).result()

if run.state.result_state != RunResultState.SUCCESS:
    raise Exception(f"job '{job_name}' run {run.run_id} finished with {run.state.result_state}")
```

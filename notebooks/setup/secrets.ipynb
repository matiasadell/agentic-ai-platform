```python
from databricks.sdk import WorkspaceClient

secret_list = [
    {
        "scope": "finsight",
        "key": "fred-api-key",
        "string_value": ""
    },
    {
        "scope": "finsight",
        "key": "news-api-key",
        "string_value": ""
    },
    {
        "scope": "finsight",
        "key": "alpha-vantage-api-key",
        "string_value": ""
    }
]

client = WorkspaceClient()

for secret in secret_list:
    if secret.get("scope") not in [s.name for s in client.secrets.list_scopes()]:
        client.secrets.create_scope(secret.get("scope"))
    client.secrets.put_secret(**secret)
```
